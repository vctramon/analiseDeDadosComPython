```python
import numpy as np
```

```python
array = np.array([1,2,3,4,5])
array 
```

```python
#1-Adição
soma = np.add(array, 10)
soma
```

```python
#2-Subtração
subtract = np.subtract(array, 5)
subtract
```

```python
#3-Multiplicação
produto = np.multiply(array, 2)
produto
```

```python
#4-Divisão
divisao = np.divide(array, 2)
divisao
```

```python
#5-Funções Trigonométricas 
seno = np.sin(array)
seno
```

```python
cosseno = np.cos(array)
cosseno
```

```python
#6-Funções
media = np.mean(array)
media
```

```python
#7-Funções de Comparação 
desvio_padra = np.std(array) 
comparacao_maior_que_3 = array > 3
comparacao_maior_que_3
```

```python
#8-Combinado Funções 
resultado = np.sqrt(array) + np.power(array, 2)
resultado 
```

```python
#9-Funções Exponenciais
exponencial = np.exp(array)
exponencial
```

```python
#10-Função de Logaritmo
logaritmo = np.log(array)
logaritmo
```

```python
arrendondado = np.round(array/ 2)
arrendondado
```

```python
#12-Função de Potência 
potencia = np.power(array, 3)
potencia 
```
