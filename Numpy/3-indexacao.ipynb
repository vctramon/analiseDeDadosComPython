```python
import numpy as np
```

```python
# Buscar elemento na linha 1 e coluna 2 

elemento = array[1,2]
elemento 
```

```python
# Pegar dados da primeira linha 
linha_0 = array [0]
linha_0
```

```python
# Acessando a coluna 1 
coluna_1 = array[:, 1]
coluna_1
```

```python
# Fatiando para obter um subarray 
#Linha 0 a 1 / Coluna 1 a 2 
subarray = array[0:2, 1:3]
subarray
```
